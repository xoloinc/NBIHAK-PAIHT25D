### Explain shortly what a figure, axes, axis and an artist is in Matplotlib.

---
**figure** = is the whole image, or figure

**axes** = primary interface of the most of the plot diagram. The axis:s, the markers. "The total" the whole plot area

**axis** =  the individual axis.. X,Y are two different Axis:s.

**artist** = the absolute top of the plot, the base that draws everything on the canvas

### Create a plot of the function y = x^2 [from -4 to 4, hint use the np.linspace function] both in the object-oriented approach and the pyplot approach. Your plot should have a title and axis-labels.

```python


x = np.linspace(-4, 4, 100) 
x2 = x**2


figure, axes = plt.subplots()
figure.suptitle('A prerry nice header') #testing around  
#axes.plot(x2, x) TODO: Why it's not working!!!!
axes.plot(x, x2) #oh... should have named it better
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('y = x²')
axes.grid(True)
plt.show()
```

### Create a figure containing 2  subplots where the first is a scatter plot and the second is a bar plot. You have the data below. 

```python
# Data for scatter plot
np.random.seed(15)
random_data_x = np.random.randn(1000)
random_data_y = np.random.randn(1000)

# Data for bar plot
fruit_data = {'grapes': 22, 'apple': 8, 'orange': 15, 'lemon': 20, 'lime': 25}
names = list(fruit_data.keys())
values = list(fruit_data.values())
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5),layout='constrained') #Just learned matplotlib uses imperial units, 2025...figsize is inches..omg
# scatter
ax1.scatter(random_data_x, random_data_y, color = "#FF000055") #röd med alpha kanal
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_title('Scatter')
ax1.grid(True)

#let's go to the Bar..s
ax2.bar(names, values, color=['purple', 'red', 'orange', 'yellow', 'lime'],edgecolor='#FF00FF33', linewidth=1.5)
ax2.set_xlabel('Fruit')
ax2.set_ylabel('Quantity')
ax2.set_title('Bars')

```
