Vad är the bias variance trade-off ? Varför är mer komplexa modeller inte
alltid bättre?

Bias = undefitting, modellen är träna på för bra data, så den klarar in varianterna
Variande = modellen är tränad på för dålig data, brus gör att den inte ser klart.

På komplexa modeller, kan den ha lärt sig detaljer, och presterar bra i tester, men när den få ny data är den dålig på det, så den får mer fel.
